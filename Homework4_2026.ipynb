#### Coriolis force

When a puck slides on a rotating turntable, it can come instantaneously to rest. From a perspective looking straight down on the turntable, plot the shape of the puck's path before it comes to rest. In a markdown cell, explain why this happens.

```python
Omega=1
R=1
d=0.2

x0=d
y0=-np.sqrt(R**2-d**2)
v=Omega*d
t_rest=-y0/v

t=np.linspace(0,t_rest,500)
eta=(x0+1j*(y0+v*t))*np.exp(-1j*Omega*t)



edge=np.linspace(0,2*np.pi,300)

plt.figure(figsize=(7,7))
plt.plot(R*np.cos(edge),R*np.sin(edge),color="gray",label="edge of turntable")
plt.plot(eta.real,eta.imag,label="path before it stops")
plt.plot(eta[-1].real,eta[-1].imag,"ro",label="at rest")
plt.plot(0,0,"k+",markersize=12,label="center")
plt.gca().set_aspect("equal")
plt.xlabel("x (m)")
plt.ylabel("y (m)")
plt.title("Seen from above, rotating with the turntable",fontsize=14)
plt.legend(fontsize=11,loc="center left",bbox_to_anchor=(1,0.5))
plt.show()
```

The puck moves in a strait line in the inertial frame. However, in the rotation frame on the turntable, the rotation of whole frame makes the strait line looks like a curve. At the end of the path, the velocity of puck equals the velocity of turn table, so it looks like rest.

#### Surface deformation due to spinning

A bucket of water is spun about its vertical axis with angular speed $\Omega$. The water in the very center of the bucket has depth $h_0$. Assume the water has settled in equilibrium relative to the bucket. Plot the height of the water surface $h - h_0$ across a diameter of the bucket. To find $h(r)$, use cylindrical polar coordinates and remember that the surface is an equipotential under the combined effects of the gravitational and centrifugal forces.

```python
g=9.8
Omega=2*np.pi
R=0.15

x=np.linspace(-R,R,300)
dh=Omega**2*x**2/(2*g)

plt.figure(figsize=(8,4))
plt.plot(x*100,dh*100)
plt.xlabel("position along diameter (cm)")
plt.ylabel(r"$h-h_0$ (cm)")
plt.title(r"$\Omega=2\pi$ rad/s, $R$=15 cm")
plt.show()
```

#### Ball drop on rotating Earth

A ball is dropped from a very tall stepladder on the equator. Plot the ball's path $z(x)$ from a tower due north of the stepladder. Explain in a markdown cell why the object lands east of its point of release.

```python
g=9.8
Omega=7.29e-5
h=100

t_land=np.sqrt(2*h/g)
t=np.linspace(0,t_land,300)
x=Omega*g*t**3/3
z=h-g*t**2/2

plt.figure(figsize=(6,6))
plt.plot(x*100,z,label="ball")
plt.plot([0,0],[0,h],"--",color="gray",label="straight down")
plt.xlabel("x, east (cm)")
plt.ylabel("z (m)")
plt.legend()
plt.show()

```

While the gravity makes the ball fall downward, the Coriolis force gives ball a force toward east, moving the ball a little bit to the east when it is landing.
